## 6. Inconsistent Value Analysis

Categorical and text-based columns are examined for inconsistent formatting, including leading or trailing whitespace, capitalization differences, and variations in category labels. The identified inconsistencies will be standardized where appropriate.

```python
text_columns = df.select_dtypes(include = "str").columns

whitespace_summary = {}

for column in text_columns:
    leading_trailing_spaces = (
        df[column].dropna().astype(str)
        != df[column].dropna().astype(str).str.strip()
    ).sum()
    
    whitespace_summary[column] = leading_trailing_spaces

whitespace_summary
```

## 7. Data Cleaning

Based on the findings from the previous analysis, the identified data quality issues are addressed while preserving legitimate information.

The following cleaning operations are performed:

- Remove leading and trailing whitespace from text fields.
- Convert empty text values into proper missing values.
- Preserve legitimate missing values in contract and salary fields.
- Maintain valid categorical and Boolean values.
- Remove any accidental duplicate or redundant columns identified during validation.

```python
text_columns = df.select_dtypes(include="str").columns

for column in text_columns: 
    df[column] = df[column].str.strip()
```

```python
#standardizing missing text values

for column in text_columns:
    df[column] = df[column].replace("", pd.NA)
```

```python
#check that whitespace issues are gone

whitespace_check = {}

for column in text_columns:
    whitespace_check[column] = (
        df[column].dropna().astype(str)
        != df[column].dropna().astype(str).str.strip()
    ).sum()

whitespace_check
```

```python
df.loc[df["company"].eq("PW"), ["title", "company"]]
```

### 7.2 Missing Value Handling

Missing values were reviewed after the text-cleaning process.

Missing values in `contract_type`, `contract_time`, and salary fields were retained because there was insufficient information to determine their actual values. Replacing these values with assumptions could introduce inaccurate information into the dataset.

The single missing company value was also investigated using the surrounding job-posting information.

```python
#check the remaining missing values

missing_after_text_cleaning = pd.DataFrame({
    "Missing Values": df.isnull().sum(),
    "Missing Percentage": (df.isnull().sum() / len(df)) * 100
})

missing_after_text_cleaning[
    missing_after_text_cleaning["Missing Values"] > 0
    ].sort_values(
        by="Missing Values",
        ascending=False
    )
```

```python
#inspect the one missing company value

df[df["company"].isna()][
    ["title", "location", "category", "description", "search_country", "country_name"]
    ]
```

```python
#check whether any empty strings remain

empty_string_counts = {}

for column in text_columns:
    empty_string_counts[column]= (df[column]=="").sum()

empty_string_counts
```
